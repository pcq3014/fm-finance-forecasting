```python
import pandas as pd

# =========================
# 1. Đọc file PMI Excel
# =========================
pmi_df = pd.read_excel("D:/fm-forecasting/data/macro/pmi.xlsx")

# =========================
# 2. Đổi tên cột
# =========================
pmi_df = pmi_df.rename(columns={
    "Năm": "year",
    "Tháng": "month",
    "Chỉ số PMI": "pmi"
})

# =========================
# 3. Tạo cột tháng
# =========================
pmi_df["month_date"] = pd.to_datetime(
    pmi_df["year"].astype(str) + "-" +
    pmi_df["month"].astype(str).str.zfill(2) + "-01"
)

# =========================
# 4. Sắp xếp và tạo PMI trễ 1 tháng
# =========================
pmi_df = pmi_df.sort_values("month_date")
pmi_df["pmi_lag1"] = pmi_df["pmi"].shift(1)

# =========================
# 5. Lưu file PMI đã xử lý
# =========================
pmi_df.to_csv("D:/fm-forecasting/data/macro/PMI_Vietnam_processed.csv", index=False, encoding="utf-8-sig")
```

### Lãi suất


```python
import pandas as pd

# =========================
# 1. Đọc file lãi suất
# =========================
rate_df = pd.read_csv("D:/fm-forecasting/data/macro/Vietnam_Refinancing_Rate.csv")

# =========================
# 2. Đổi tên cột
# =========================
rate_df = rate_df.rename(columns={
    "Year": "year",
    "Date (YYYY-MM)": "month",
    "Rate (%)": "interest_rate"
})

# =========================
# 3. Tạo cột month_date
# =========================
rate_df["month_date"] = pd.to_datetime(rate_df["month"] + "-01")

# =========================
# 4. Sắp xếp
# =========================
rate_df = rate_df.sort_values("month_date")

# =========================
# 5. Tạo thêm biến trễ nếu cần
# =========================
rate_df["interest_rate_lag1"] = rate_df["interest_rate"].shift(1)

# =========================
# 6. Lưu lại file đã xử lý
# =========================
rate_df.to_csv("interest_rate_processed.csv", index=False, encoding="utf-8-sig")
```
