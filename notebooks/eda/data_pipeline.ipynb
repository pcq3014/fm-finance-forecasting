### Thực hiện kiểm tra dữ liệu cơ bản

```python
import pandas as pd
import os
import glob

# Đọc và gộp dữ liệu 
folder_path = 'data/raw/v2' 
csv_files = glob.glob(os.path.join(folder_path, "*.csv"))
df_list = []
for file in csv_files:
    symbol = os.path.basename(file).split('.')[0].upper()
    if len(symbol) <= 4: 
        df = pd.read_csv(file, parse_dates=['Date'])
        df['Symbol'] = symbol
        df_list.append(df)

data = pd.concat(df_list)

# ==========================================
# THỰC HIỆN EDA CƠ BẢN 
# ==========================================

print("="*50)
print("BÁO CÁO TỔNG QUAN DỮ LIỆU VN30")
print("="*50)

# 1. Thống kê chung toàn thị trường (Toàn bộ Dataset)
print("[1] THỐNG KÊ CHUNG:")
print(f"    - Tổng số mã cổ phiếu: {data['Symbol'].nunique()} mã")
print(f"    - Tổng số dòng dữ liệu : {len(data):,} dòng")
print(f"    - Ngày bắt đầu (Min) : {data['Date'].min().strftime('%Y-%m-%d')}")
print(f"    - Ngày kết thúc (Max): {data['Date'].max().strftime('%Y-%m-%d')}")
print(f"    - Tổng số dữ liệu trống (Missing): {data.isnull().sum().sum()} ô")
print("-" * 50)

# 2. Thống kê chi tiết cho từng mã cổ phiếu
print("[2] CHI TIẾT THEO TỪNG MÃ:")

# Nhóm dữ liệu theo mã cổ phiếu và tổng hợp thông tin
summary_df = data.groupby('Symbol').agg(
    Số_dòng=('Date', 'count'),
    Ngày_bắt_đầu=('Date', lambda x: x.min().strftime('%Y-%m-%d')),
    Ngày_kết_thúc=('Date', lambda x: x.max().strftime('%Y-%m-%d')),
    Giá_trị_trống=('Close', lambda x: x.isnull().sum())
)

# Hiển thị bảng tổng hợp
print(summary_df.to_string())
print("="*50)
```

```python
# Lọc ra các mã có ngày bắt đầu từ 2013-01-02
filtered_symbols = summary_df[summary_df['Ngày_bắt_đầu'] == '2013-01-02']

print("Các mã cổ phiếu có ngày bắt đầu từ 2013-01-02:")
print(filtered_symbols)

# Lấy danh sách các mã
symbols_2013 = filtered_symbols.index.tolist()
print("\nDanh sách mã:", symbols_2013)
```
