```python
import pandas_datareader.data as web
import datetime

def get_worldbank_data(country_code='VN'):
    start = datetime.datetime(2013, 1, 1)
    end = datetime.datetime(2026, 3, 25)
    
    # NY.GDP.MKTP.CD: GDP (current US$)
    # BX.KLT.DINV.CD.WD: FDI (Foreign direct investment, net inflows)
    indicators = {'NY.GDP.MKTP.CD': 'GDP', 'BX.KLT.DINV.CD.WD': 'FDI'}
    
    df = web.get_data_worldbank(indicators, country=country_code, start=start, end=end)
    return df
```

```python
import wbgapi as wb
import pandas as pd

indicators = {
    'NY.GDP.MKTP.CD': 'GDP_Nominal_USD',
    'NY.GDP.MKTP.KD': 'GDP_Real_USD',
    'NY.GDP.MKTP.ZG': 'GDP_Growth_Pct'
}

# 1. Lấy dữ liệu
df_vn = wb.data.DataFrame(list(indicators.keys()), 'VNM', mrv=13)

# 2. XOAY BẢNG TRƯỚC (Để Năm nhảy về làm Index)
df_vn = df_vn.T 

# 3. BÂY GIỜ MỚI XỬ LÝ INDEX (Vì lúc này Index là ['YR2013', 'YR2014',...])
df_vn.index = df_vn.index.str.replace('YR', '').astype(int)

# 4. Đổi tên cột
df_vn.columns = [indicators.get(col, col) for col in df_vn.columns]
df_vn = df_vn.sort_index()

print("--- Kết quả sau khi fix lỗi ---")
print(df_vn.head())
```
