```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Khai báo danh sách các file kết quả
files = {
    'ARIMA': 'arima_summary.csv',
    'LSTM': 'lstm_close_summary.csv',
    'Chronos': 'chronos_summary.csv',
    'TimesFM': 'timesfm_summary.csv',
    'Moirai': 'moirai_summary.csv'
}

# 2. Đọc và gộp dữ liệu
all_data = []
overall_data = []

for model_name, file_path in files.items():
    try:
        df = pd.read_csv(file_path)
        df['Model'] = model_name
        
        # Tách riêng dòng OVERALL và các dòng Ticker
        overall_row = df[df['ticker'] == 'OVERALL'].copy()
        ticker_rows = df[df['ticker'] != 'OVERALL'].copy()
        
        overall_data.append(overall_row)
        all_data.append(ticker_rows)
    except Exception as e:
        print(f"Error loading {file_path}: {e}")

df_overall = pd.concat(overall_data, ignore_index=True)
df_tickers = pd.concat(all_data, ignore_index=True)

# Thiết lập phong cách biểu đồ
sns.set_theme(style="whitegrid")
```

```python
# =====================================================================
# BIỂU ĐỒ 1: SO SÁNH CHỈ SỐ TỔNG THỂ (OVERALL BAR CHARTS)
# =====================================================================
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Overall Performance Comparison Across Models', fontsize=16, weight='bold')

metrics = ['MAE', 'RMSE', 'DA', 'IC']
titles = ['Mean Absolute Error (MAE) ↓', 'Root Mean Squared Error (RMSE) ↓', 
          'Directional Accuracy (DA) ↑', 'Information Coefficient (IC) ↑']

for i, ax in enumerate(axes.flatten()):
    metric = metrics[i]
    # Sắp xếp để biểu đồ cột trực quan hơn
    sorted_df = df_overall.sort_values(by=metric, ascending=(metric in ['MAE', 'RMSE']))
    
    sns.barplot(data=sorted_df, x='Model', y=metric, ax=ax, palette='viridis')
    ax.set_title(titles[i], fontsize=12)
    ax.set_xlabel('')
    ax.set_ylabel(metric)
    
    # Ghi giá trị trực tiếp lên đầu các cột
    for p in ax.patches:
        ax.annotate(f"{p.get_height():.3f}", 
                    (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha='center', va='bottom', fontsize=10, xytext=(0, 5), 
                    textcoords='offset points')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig('overall_comparison.png', dpi=300, bbox_inches='tight')
plt.close()
```

```python
# =====================================================================
# BIỂU ĐỒ 2: PHÂN PHỐI CHỈ SỐ DA VÀ IC TRÊN 15 TICKER (BOXPLOTS)
# =====================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Distribution of Metrics Across 15 Tickers', fontsize=16, weight='bold')

# Biểu đồ phân phối DA
model_order_da = df_tickers.groupby('Model')['DA'].median().sort_values(ascending=False).index
sns.boxplot(data=df_tickers, x='Model', y='DA', ax=axes[0], order=model_order_da, palette='Set2')
sns.stripplot(data=df_tickers, x='Model', y='DA', ax=axes[0], order=model_order_da, color='black', alpha=0.5, size=4)
axes[0].set_title('Directional Accuracy (DA) Distribution', fontsize=12)
axes[0].set_ylabel('DA')
axes[0].set_xlabel('')

# Biểu đồ phân phối IC
model_order_ic = df_tickers.groupby('Model')['IC'].median().sort_values(ascending=False).index
sns.boxplot(data=df_tickers, x='Model', y='IC', ax=axes[1], order=model_order_ic, palette='Set2')
sns.stripplot(data=df_tickers, x='Model', y='IC', ax=axes[1], order=model_order_ic, color='black', alpha=0.5, size=4)
axes[1].set_title('Information Coefficient (IC) Distribution', fontsize=12)
axes[1].set_ylabel('IC')
axes[1].set_xlabel('')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig('ticker_distribution.png', dpi=300, bbox_inches='tight')
plt.close()
```

```python
# =====================================================================
# BIỂU ĐỒ 3: BIỂU ĐỒ RADAR SO SÁNH ĐA CHIỀU
# =====================================================================
def normalize(df, cols, invert=[]):
    """Chuẩn hóa dữ liệu về [0, 1]. Invert các chỉ số mà 'càng nhỏ càng tốt'."""
    norm_df = df.copy()
    for col in cols:
        min_val = df[col].min()
        max_val = df[col].max()
        if max_val == min_val:
            norm_df[col] = 1.0
        else:
            norm_df[col] = (df[col] - min_val) / (max_val - min_val)
        if col in invert:
            norm_df[col] = 1 - norm_df[col]
    return norm_df

norm_overall = normalize(df_overall, ['MAE', 'RMSE', 'DA', 'IC'], invert=['MAE', 'RMSE'])

categories = ['MAE (Inverted)', 'RMSE (Inverted)', 'DA', 'IC']
N = len(categories)
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
plt.xticks(angles[:-1], categories, fontsize=12)
ax.set_rlabel_position(0)
plt.yticks([0.25, 0.5, 0.75], ["0.25", "0.50", "0.75"], color="grey", size=10)
plt.ylim(0, 1)

colors_radar = plt.cm.Set1.colors
for i, row in norm_overall.iterrows():
    values = row[['MAE', 'RMSE', 'DA', 'IC']].tolist()
    values += values[:1]
    ax.plot(angles, values, linewidth=2, linestyle='solid', label=row['Model'], color=colors_radar[i%len(colors_radar)])
    ax.fill(angles, values, alpha=0.1, color=colors_radar[i%len(colors_radar)])

plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.title('Overall Performance Radar (Vòng càng ngoài càng tốt)', size=15, weight='bold', y=1.1)

plt.savefig('radar_chart.png', dpi=300, bbox_inches='tight')
plt.close()

print("Hoàn thành tạo 3 biểu đồ: overall_comparison.png, ticker_distribution.png, radar_chart.png")
```
