```python
from vnstock import Quote

start = "2013-01-01"
end = "2026-05-12"

# VN-Index
vnindex = Quote(symbol="VNINDEX", source="VCI").history(
    start=start,
    end=end,
    interval="1D"
)

# VN30
vn30 = Quote(symbol="VN30", source="VCI").history(
    start=start,
    end=end,
    interval="1D"
)

print(vnindex.head())
print(vn30.head())
```

```python
vnindex.to_csv("vnindex_2013_2026.csv", index=False)
vn30.to_csv("vn30_2013_2026.csv", index=False)
```
